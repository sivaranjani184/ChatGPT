```python
from transformers import pipeline

# Use the exact string "summarization"
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# BART requires longer text to summarize properly
text = "Python is a high-level, general-purpose programming language. Its design philosophy emphasizes code readability with the use of significant indentation. It is dynamically typed and garbage-collected."

# Add length constraints to prevent errors with short text
summary = summarizer(text, max_length=50, min_length=20, do_sample=False)

print(summary)
```
